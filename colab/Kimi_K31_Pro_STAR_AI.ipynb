# STAR AI — Kimi K3.1 Pro on Google Colab

هذا الدفتر يرسل الطلبات عبر **Kimi API**، ولا يحتاج GPU أو تنزيل أوزان النموذج. لا تحفظ المفتاح داخل الدفتر ولا تشاركه.

1. افتح [platform.kimi.ai](https://platform.kimi.ai/) وأنشئ مفتاح API شخصيًا.
2. من Colab > Secrets أضف `MOONSHOT_API_KEY`، أو أدخله عند التشغيل بواسطة مطالبة مخفية.
3. نفّذ الخلايا بالترتيب. استخدام API قد يخضع لتكاليف منفصلة عن اشتراك Colab Pro.

```python
from getpass import getpass
from openai import OpenAI

api_key = None
try:
    from google.colab import userdata
    api_key = userdata.get('MOONSHOT_API_KEY')
except (ImportError, Exception):
    pass
if not api_key:
    api_key = getpass('Kimi API key (hidden): ')
if not api_key:
    raise ValueError('MOONSHOT_API_KEY is required')

client = OpenAI(api_key=api_key, base_url='https://api.moonshot.ai/v1', timeout=90.0, max_retries=1)
MODEL = 'kimi-k3.1-pro'
print('Client ready. No key displayed.')
```

```python
question = 'Explain a secure architecture for integrating an AI coding assistant with a Linux terminal, requiring approval before any command execution.'

try:
    result = client.chat.completions.create(
        model=MODEL,
        messages=[
            {'role': 'system', 'content': 'You are STAR AI. Answer in Iraqi Arabic. Provide safe, accurate software engineering guidance; do not claim to have executed commands.'},
            {'role': 'user', 'content': question},
        ],
        reasoning_effort='high',
        max_tokens=1200,
    )
    print(result.choices[0].message.content or 'No text returned')
except Exception as exc:
    print('Kimi request failed:', type(exc).__name__)
    print('Check the API key, provider balance, model availability and network connectivity. Do not paste keys into chat.')
```
